```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
vendas = pd.read_csv('vendas.csv')
df_clientes = pd.read_csv('clientes.csv')   
df_produtos = pd.read_csv('produtos.csv')
df_campanhas = pd.read_csv('campanhas.csv')
df_avaliacoes = pd.read_csv('avaliacoes.csv')
df_atendimentos = pd.read_csv('atendimentos.csv')
vendas1 = pd.read_csv('venda 1.csv')
vendas2 = pd.read_csv('venda 2.csv' )
vendas3 = pd.read_csv('venda 3.csv' )
```

```python
df_vendas = pd.concat([vendas, vendas1, vendas2, vendas3], ignore_index=True)
```

```python
df_clientes.columns = df_clientes.columns.str.strip().str.lower()
df_vendas.columns = df_vendas.columns.str.strip().str.lower()
df_produtos.columns = df_produtos.columns.str.strip().str.lower()
```

```python
clientes_vendas = pd.merge(df_vendas, df_clientes, on="id_cliente", how="left")
tabela_final = pd.merge(clientes_vendas, df_produtos, on="id_produto", how="left")
```

```python
tabela_final.groupby("nome_produto")["preco"].sum().reset_index().sort_values(by="preco", ascending=False)
```

```python
tabela_produtos = tabela_final[['nome_produto', 'preco']].drop_duplicates().reset_index(drop=True)
tabela_produtos = tabela_produtos.dropna()
tabela_produtos
```

```python
tabela_produtos_array = np.array(tabela_produtos['preco'])

q1 = np.percentile(tabela_produtos_array, 25)
q2 = np.percentile(tabela_produtos_array, 50)
q3 = np.percentile(tabela_produtos_array, 75)
mediana = np.median(tabela_produtos_array) # median = median
media = np.mean(tabela_produtos_array) # mean = média

print(f'Primeiro quartil (Q1): {q1}')
print(f'Primeiro quartil (Q2): {q2}')
print(f'Primeiro quartil (Q3): {q3}')
print(f'Mediana {mediana}')
print(f'Média {media}')
```

```python
contagem = tabela_final['nome_produto'].value_counts().reset_index()
contagem
```

```python
count_array = np.array(contagem['count'])

q1 = np.percentile(count_array, 25)
q2 = np.percentile(count_array, 50)
q3 = np.percentile(count_array, 75)
mediana = np.median(count_array) # median = median
media = np.mean(count_array) # mean = média

distancia =  (media - mediana) / mediana
iqr = q3 - q1
limite_inferior = q1 - (1.5 * iqr)

print(f'Primeiro quartil (Q1): {q1}')
print(f'Primeiro quartil (Q2): {q2}')
print(f'Primeiro quartil (Q3): {q3}')
print(f'Mediana {mediana}')
print(f'Média {media}')
print(f'Limite Inferior: {limite_inferior}')
print(f'Distância: {distancia}')
```

```python
count_outliers = contagem[contagem['count'] < media]
count_outliers.tail()
```
